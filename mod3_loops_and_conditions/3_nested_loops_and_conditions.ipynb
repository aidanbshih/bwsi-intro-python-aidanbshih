```python
for number in range(1, 101):
    if number%3 == 0 and number%5 == 0: # check if the number is a multiple of both 3 and 5.
        print("FizzBuzz")
        continue                        # if so, we don't need to run through any of the other code.
    if number%3 == 0:                   # check if the number is only a multiple of 3.
        print("Fizz")                   
        continue                        # we can skip the rest of the code becasue we already know it's not a multiple of 5.
    if number%5 == 0:                   # check if the number is only a multiple of 5.
        print("Buzz")
        continue                        # we don't want to print the number if it's a multiple of 5.
    print(number)                       # only runs if the number is not a multiple of 3 nor of 5.
```

### Q.3 Please complete the commented lines below with descriptions of the code as indicated. 

```python
# Create list to store prime numbers
prime_numbers_list = []

# This for loop iterates over every number 2-10.
for n in range(2,11):
    # Set is_prime as True
    is_prime = True
    # This for loop iterates over every number from 2-the current n value.
    for k in range(2, n):
        # This condition checks if n is a multiple of k (if n is divisible by k)
        if n % k == 0:
            # If n % k == 0, then n is a multiple of k, and n is not prime. Print the whole number result of dividing n by k.
            print(f"{n} = {k} * {n // k}")
            is_prime = False
            break
    # This condition checks if is_prime is True
    if is_prime:
        # If n is prime, then print as output 
        # and append to list of prime numbers
        print(f"{n} is a prime number")
        prime_numbers_list.append(n)
        
print(f"\nList of prime numbers in range: {prime_numbers_list}")
```
